```python
#a function that has (while true loop) so that a user can only input (yes or no) for example
#(I didn't end up using this and I don't know if it works, but it is a good idea)
'''
while True:
  if VARIABLE != "yes" or VARIABLE != "no":
    print("That is not a valid input")
  else:
   break
'''
#This bit here may seem inefficient, but I prefer copy and pasting "print("")" over having one long line of code with a bunch of "\n"s
#It allows me to see all the directions without having to scroll 3 miles to the end of the line.
print("This is a snowman game. Your objective is to guess the secret word.")
print("There will be a number of dashes equal to the length of your word, and each correct guess will transform those dashes into the correct letter.")
print("You may choose your game difficulty and your word difficulty. There are two word banks that you can choose from.\n")
print("When you choose your game difficulty:")
print("-Easy gives you a number of guesses equal to the length of the secret word plus 5")
print("-Medium gives you a number of guesses equal to the length of the secret word plus 3")
print("-Hard gives you a number of guesses equal to the length of the secret word plus 1\n")
print("The game ends when either you guess the word or run out of guesses.")
print("Once you run out of guesses, you will have one final opportunity to guess the secret word.\n")
print("At the very end of the game, it will tell you how many points you obtained.")
print("You get a different number of points per correct guess depending on the game difficulty.\n")
print("Good luck!")

points = 0
points_per_guess = 0
easy_word_bank = ["entrepreneur", "calculus", "jazz", "embargo", "rebuking", "wounding", "chiefdom", "kimono"]
hard_word_bank = ["hajj", "syzygy", "rhythm", "buzzy", "bouillon", "velleity"]
import random
wordbankchoice = input("Would you like to use the harder word bank? (input yes if yes)")
#I used google to figure out how to use the random library
if wordbankchoice == "yes":
  secret_word = random.choice(hard_word_bank)
  print(secret_word)
else:
  secret_word = random.choice(easy_word_bank)
  print(secret_word)

hidden_word = []
for i in range(len(secret_word)):
  hidden_word.append("-")
print(" ".join(hidden_word))

#I used chatGPT here to help me figure out why guesses wasn't being assigned a value.
#It basically told me I needed to add "global" stuff whih we haven't learned, so I asked it for a different way and it wasn't helping
#So I essentially just experimented for a while and then I thought of just making "guesses" a parameter and that ended up working
#I also added a bunch of other variables as parameters, and thats probably pretty inefficient but it works for me.
def choose_difficulty(difficulty, points_per_guess, guesses):
  if difficulty == "easy":
    guesses = len(secret_word) + 5
    points_per_guess = 1
  elif difficulty == "medium":
    guesses = len(secret_word) + 3
    points_per_guess = 2
#Little bit of an easter egg here: if the user types "override" as their game difficulty, it allows them to choose the number of guesses they get
#But it also makes their points per guess 0.
#this is not an extra feature, I just added it for fun
  elif difficulty == "override":
#Here is my TRY/EXCEPT
    try:
      guesses = int(input("How many guesses would you like?"))
    except ValueError:
                    print("Please enter a valid number!")
    points_per_guess = 0
  else:
    guesses = len(secret_word) + 1
    points_per_guess = 3
#I never knew you could return multiple variables (or maybe we learned it in class, and I just forgot)
#so I looked up if you could and the answer ended up being yes
  return guesses, points_per_guess


def guess(letter, guesses, secret_word, hidden_word, points, points_per_guess):
  guesses -= 1
#I did't realize it before running my code a few times to try to break it, but the !="" covers all the different spaces
#(spaces, tabs, returns, etc)
  if letter in secret_word and letter != "":
    print(f"Your guess was correct! {letter} appears in the secret word.")
    for i in range(len(secret_word)):
      if secret_word[i] == letter:
        hidden_word[i] = letter
    points += points_per_guess
  else:
    print(f"Your guess was incorrect. {letter} doesn't appear in the secret word.")
  return guesses, hidden_word, points

guesses = 0
difficulty = (input("What difficulty would you like? (easy, medium or hard)\n")).lower()
#I don't know fi the order of "guesses, points_per_guess = ..." matters, but the return inside the function follow that order so I just figured it does matter.
guesses, points_per_guess = choose_difficulty(difficulty, points_per_guess, guesses)
print(f"Number of guesses: {guesses}")
print(f"Points per guess: {points_per_guess}")


while guesses > 0:

#I could make final_guessYN a function, but I didn't feel like it (this is the beginning of the final_guessYN section)
#It also include the hint stuff
  final_guessYN = (input("Would you like to make your final guess of the secret word? Enter 'yes' if yes, enter 'no' if no\n Enter 'hint' if you would like a hint (this reveals the first two letters but costs 2 guesses) \n (some of the revealed letters may repeat later in the word)"))
  if final_guessYN == "yes":
    final_guess = input("Input your final guess\n")
    if final_guess == secret_word:
      print("Congratulations! You have fully guessed the secret word!")
      break
    else:
      print(f"You failed. \nThe secret word was {secret_word}.")
      break
  elif final_guessYN == "hint" and guesses >= 2:
    hidden_word[0:2:] = secret_word[0:2:]
    guesses -= 2
    print(f"So far, you have revealed: {" ".join(hidden_word)}\n")
#I added the following line of code because I noticed that you could still get a hint even if you didn't have 2 or more guesses left
  elif final_guessYN == "hint" and guesses < 2:
    print("You don't have enough guesses left for a hint.")
  elif final_guessYN == "no":
    print("Continuing \n")
#This is the end of the final_guessYN section

  letter = str((input("Guess a letter\n"))).lower()
  guesses, hidden_word, points = guess(letter, guesses, secret_word, hidden_word, points, points_per_guess)
#I used AI to teach me how to debug this line of code, because there was a problem between the .join and the "hidden_list" list
  if "".join(hidden_word) == secret_word:
    print("Congratulations! You have fully guessed the secret word!")
    points += 10
    break
  print(f"So far, you have revealed: {" ".join(hidden_word)}\n")
  print(f"You have {guesses} guesses left.")
  print(f"You have {points} points.")

if guesses == 0 and "".join(hidden_word) != secret_word:
  final_guess = input("Input your final guess\n")
  if final_guess == secret_word:
      print("Congratulations! You have fully guessed the secret word!")
      points += 10
  else:
    print(f"You failed. \nThe secret word was {secret_word}.")

print(f"You ended with {points} points.")
```
